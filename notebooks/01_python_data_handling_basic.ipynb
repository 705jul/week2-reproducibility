```python
before_rows = len(df_clean)
df_clean = df_clean.drop_duplicates().copy()
after_rows = len(df_clean)
print("중복 제거 전 행 수:", before_rows)
print("중복 제거 후 행 수:", after_rows)
print("제거된 행 수:", before_rows - after_rows)
```

```python
df_clean.isnull().sum()
```

```python
age_median = df_clean["age"].median()
monthly_fee_median = df_clean["monthly_fee"].median()
df_clean["age"] = df_clean["age"].fillna(age_median)
df_clean["monthly_fee"] = df_clean["monthly_fee"].fillna(monthly_fee_median)
print("age median:", age_median)
print("monthly_fee median:", monthly_fee_median)
```

```python
df_clean["gender"] = df_clean["gender"].fillna("unknown")
```

```python
df_clean["contract_type"] = df_clean["contract_type"].fillna("unknown")
```

```python
df_clean.isnull().sum()
```

```python
selected_cols = [
    "customer_id",
    "age",
    "gender",
    "monthly_fee",
    "contract_type",
    "usage_days",
    "support_calls",
    "churn"
]
df_selected = df_clean[selected_cols]
df_selected.head()
```

```python
usage_over_30 = df_clean[df_clean["usage_days"] >= 30]
usage_over_30.head()
```

```python
high_fee_customers = df_clean[df_clean["monthly_fee"] >= 50000]
high_fee_customers.head()
```

```python
churn_customers = df_clean[df_clean["churn"] == 1]
churn_customers.head()
```

```python
df_clean["is_high_fee"] = (df_clean["monthly_fee"] >= 70000).astype(int)
df_clean[["customer_id", "monthly_fee", "is_high_fee"]].head()
```

```python
def make_usage_group(days):
    if days < 90:
        return "short"
    elif days < 365:
        return "middle"
    else:
        return "long"
df_clean["usage_group"] = df_clean["usage_days"].apply(make_usage_group)
df_clean[["customer_id", "usage_days", "usage_group"]].head()
```

```python
summary_fee = (
    df_clean
    .groupby("contract_type")["monthly_fee"]
    .mean()
    .reset_index()
)
summary_fee
```

```python
summary_churn = (
    df_clean
    .groupby("contract_type")["churn"]
    .mean()
    .reset_index()
)
summary_churn = summary_churn.rename(columns={
    "churn": "churn_rate"
})
summary_churn
```

```python
usage_churn = (
    df_clean
    .groupby("usage_group")["churn"]
    .mean()
    .reset_index()
)
usage_churn = usage_churn.rename(columns={
    "churn": "churn_rate"
})
usage_churn["churn_rate_percent"] = usage_churn["churn_rate"] * 100
usage_churn
```

```python
contract_summary = (
    df_clean
    .groupby("contract_type")
    .agg(
        customer_count=("customer_id", "count"),
        avg_monthly_fee=("monthly_fee", "mean"),
        churn_rate=("churn", "mean")
    )
    .reset_index()
)
contract_summary
```
