````python
from pathlib import Path
import os
import sys
import json
import textwrap


# 현재 실행 위치가 notebook 또는 notebooks 폴더이면 부모 폴더를 프로젝트 루트로 사용
BASE_DIR = Path.cwd().parent if Path.cwd().name in ["notebook", "notebooks"] else Path.cwd()
BASE_DIR = BASE_DIR.resolve()

# 폴더 준비
REPORT_DIR = BASE_DIR / "reports" / "week2"
OUTPUT_DIR = BASE_DIR / "outputs"

REPORT_DIR.mkdir(parents=True, exist_ok=True)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


# 기본 설정값
AUTHOR = globals().get(
    "AUTHOR",
    globals().get("author", os.environ.get("AUTHOR", "student"))
)

AWS_REGION = globals().get(
    "AWS_REGION",
    os.environ.get("AWS_REGION", "ap-northeast-2")
)

S3_BUCKET = globals().get(
    "S3_BUCKET",
    globals().get("BUCKET", globals().get("bucket_name", os.environ.get("S3_BUCKET", "edu-ai-lake")))
)

S3_CLEAN_PREFIX = globals().get(
    "S3_CLEAN_PREFIX",
    os.environ.get("S3_CLEAN_PREFIX", f"users/{AUTHOR}/clean/customers/")
)

S3_CLEAN_KEY = globals().get(
    "S3_CLEAN_KEY",
    os.environ.get("S3_CLEAN_KEY", f"{S3_CLEAN_PREFIX}customers_clean_20260606_v2.csv")
)

S3_CLEAN_URI = f"s3://{S3_BUCKET}/{S3_CLEAN_KEY}"

TARGET_COLUMN = globals().get(
    "TARGET_COLUMN",
    os.environ.get("TARGET_COLUMN", "churn")
)

LOCAL_CLEAN_FILE_RAW = globals().get(
    "LOCAL_CLEAN_FILE",
    os.environ.get("LOCAL_CLEAN_FILE", "data/clean/customers_clean_20260606_v2.csv")
)

LOCAL_CLEAN_FILE = Path(LOCAL_CLEAN_FILE_RAW)

if not LOCAL_CLEAN_FILE.is_absolute():
    LOCAL_CLEAN_FILE = BASE_DIR / LOCAL_CLEAN_FILE

expected_columns = globals().get(
    "expected_columns",
    [
        "customer_id",
        "age",
        "gender",
        "monthly_fee",
        "contract_type",
        "usage_days",
        "support_calls",
        "churn",
    ]
)


def safe_relative_path(path: Path, base_dir: Path) -> str:
    """
    path가 base_dir 하위이면 상대경로를 반환하고,
    아니면 절대경로 문자열을 반환한다.
    """
    path = Path(path)

    try:
        return str(path.relative_to(base_dir))
    except ValueError:
        return str(path)


def markdown_safe(value) -> str:
    """
    Markdown 표 안에서 깨질 수 있는 값을 안전한 문자열로 변환한다.
    """
    if value is None:
        return "확인 필요"

    text = str(value).replace("\n", " ").strip()

    if not text:
        return "확인 필요"

    return text.replace("|", "\\|")


def bullet_list(items, empty_text: str = "- 없음") -> str:
    """
    리스트를 Markdown bullet 문자열로 변환한다.
    """
    if not items:
        return empty_text

    return "\n".join([f"- {item}" for item in items])


def dict_to_bullets(data: dict, empty_text: str = "- 없음") -> str:
    """
    dict를 Markdown bullet 문자열로 변환한다.
    """
    if not data:
        return empty_text

    return "\n".join([f"- {key}: {value}" for key, value in data.items()])


def safe_read_json(path: Path) -> dict:
    """
    JSON 파일을 안전하게 읽는다.
    """
    if not path.exists():
        return {}

    try:
        return json.loads(path.read_text(encoding="utf-8"))
    except json.JSONDecodeError:
        return {}


# validation_result가 있으면 사용하고, 없으면 JSON 또는 src.preprocess로 생성 시도
if "validation_result" in globals():
    validation_result_for_record = validation_result

else:
    data_validation_json_file = OUTPUT_DIR / "data_validation_report.json"
    validation_result_for_record = safe_read_json(data_validation_json_file)

    if not validation_result_for_record and LOCAL_CLEAN_FILE.exists():
        if str(BASE_DIR) not in sys.path:
            sys.path.append(str(BASE_DIR))

        try:
            from src.preprocess import read_clean_data, validate_customer_schema

            df_clean_for_record = read_clean_data(LOCAL_CLEAN_FILE)

            validation_result_for_record = validate_customer_schema(
                df=df_clean_for_record,
                expected_columns=expected_columns,
                target_column=TARGET_COLUMN,
            )

        except Exception as e:
            validation_result_for_record = {
                "row_count": "확인 필요",
                "column_count": "확인 필요",
                "error": f"{type(e).__name__}: {e}",
            }

    if not validation_result_for_record:
        validation_result_for_record = {
            "row_count": "확인 필요",
            "column_count": "확인 필요",
            "columns": {
                "expected_columns": expected_columns,
                "actual_columns": [],
                "missing_columns": [],
                "extra_columns": [],
            },
            "missing_values": {
                "columns_with_missing": [],
                "has_missing_values": "확인 필요",
            },
            "dtypes": {
                "numeric_columns": [],
                "object_columns": [],
                "date_candidate_columns": [],
                "id_candidate_columns": [],
            },
            "target": {
                "target_column": TARGET_COLUMN,
                "target_exists": "확인 필요",
                "target_missing_count": "확인 필요",
                "target_unique_count": "확인 필요",
                "target_distribution_count": {},
            },
            "leakage": {
                "possible_leakage_columns": [],
                "possible_leakage_columns_count": 0,
            },
            "baseline_ready": "확인 필요",
        }


# 검증 결과 꺼내기
columns_result = validation_result_for_record.get("columns", {})
missing_values_result = validation_result_for_record.get("missing_values", {})
dtypes_result = validation_result_for_record.get("dtypes", {})
target_result = validation_result_for_record.get("target", {})
leakage_result = validation_result_for_record.get("leakage", {})

row_count = validation_result_for_record.get("row_count", "확인 필요")
column_count = validation_result_for_record.get("column_count", "확인 필요")

expected_columns_for_report = columns_result.get("expected_columns", expected_columns)
actual_columns_for_report = columns_result.get("actual_columns", [])
missing_columns_for_report = columns_result.get("missing_columns", [])
extra_columns_for_report = columns_result.get("extra_columns", [])

columns_with_missing = missing_values_result.get("columns_with_missing", [])

missing_column_lines = [
    f"{item.get('column')}: {item.get('missing_count')}개 ({item.get('missing_ratio_percent')}%)"
    for item in columns_with_missing
]

high_missing_column_lines = [
    f"{item.get('column')}: {item.get('missing_ratio_percent')}%"
    for item in columns_with_missing
    if float(item.get("missing_ratio_percent", 0)) >= 30
]

numeric_columns = dtypes_result.get("numeric_columns", [])
object_columns = dtypes_result.get("object_columns", [])
date_candidate_columns = dtypes_result.get("date_candidate_columns", [])
id_candidate_columns = dtypes_result.get("id_candidate_columns", [])

target_distribution_count = target_result.get("target_distribution_count", {})
possible_leakage_columns = leakage_result.get("possible_leakage_columns", [])

target_like_columns = [
    col for col in possible_leakage_columns
    if TARGET_COLUMN.lower() in col.lower()
    or "churn" in col.lower()
    or "label" in col.lower()
    or "target" in col.lower()
]

future_info_columns = [
    col for col in possible_leakage_columns
    if "future" in col.lower()
    or "after" in col.lower()
    or "next" in col.lower()
    or "result" in col.lower()
    or "outcome" in col.lower()
]

review_columns = sorted(
    set(possible_leakage_columns + id_candidate_columns)
)

baseline_ready = validation_result_for_record.get("baseline_ready", "확인 필요")

baseline_reason_items = []

if baseline_ready is True:
    baseline_reason_items.append("필수 컬럼, target, 데이터 크기 조건을 1차로 만족했다.")
elif baseline_ready is False:
    if missing_columns_for_report:
        baseline_reason_items.append("필수 컬럼 누락이 있어 schema 수정이 필요하다.")
    if target_result.get("target_exists") is False:
        baseline_reason_items.append("target 컬럼이 존재하지 않는다.")
    if target_result.get("target_missing_count") not in [0, "0", None, "확인 필요"]:
        baseline_reason_items.append("target 컬럼에 결측치가 있다.")
    if target_result.get("target_unique_count") in [0, 1, "0", "1"]:
        baseline_reason_items.append("target class가 2개 미만이다.")
    if not baseline_reason_items:
        baseline_reason_items.append("검증 항목 중 일부가 통과하지 못했다.")
else:
    baseline_reason_items.append("validation_result를 먼저 생성하거나 데이터 검증을 다시 실행해야 한다.")

additional_check_items = [
    "extra columns 중 leakage 후보가 있는지 확인한다.",
    "식별자 컬럼은 feature에서 제외할지 검토한다.",
    "target 컬럼은 feature에 넣지 않는다.",
]

local_clean_file_text = safe_relative_path(LOCAL_CLEAN_FILE, BASE_DIR)

# 저장 파일 경로
period3_record_file = REPORT_DIR / "day5_period3_data_validation_record.md"

# Markdown 기록 양식
period3_record_text = f"""
# 2주차 Day 5 3교시 실습 기록지

## 작성자

- author: {AUTHOR}

## 1. 오늘 주제

데이터 검증 + 전처리 로직 재사용

## 2. 입력 데이터

- local clean file: {local_clean_file_text}
- S3 bucket: {S3_BUCKET}
- S3 key: {S3_CLEAN_KEY}
- S3 URI: {S3_CLEAN_URI}

## 3. 데이터 크기

- rows: {row_count}
- columns: {column_count}

## 4. 컬럼 검증

### expected columns

{bullet_list(expected_columns_for_report)}

### actual columns

{bullet_list(actual_columns_for_report)}

### missing columns

{bullet_list(missing_columns_for_report)}

### extra columns

{bullet_list(extra_columns_for_report)}

## 5. 결측치 확인

- 결측치가 있는 컬럼:

{bullet_list(missing_column_lines, "- 결측치가 있는 컬럼 없음")}

- 결측치 비율이 높은 컬럼:

{bullet_list(high_missing_column_lines, "- 결측치 비율이 높은 컬럼 없음")}

- 처리 필요 여부: {markdown_safe(missing_values_result.get("has_missing_values", "확인 필요"))}

## 6. dtype 확인

- 수치형 컬럼:

{bullet_list(numeric_columns)}

- 범주형 또는 object 컬럼:

{bullet_list(object_columns)}

- 날짜형 후보 컬럼:

{bullet_list(date_candidate_columns)}

- 식별자 후보 컬럼:

{bullet_list(id_candidate_columns)}

## 7. target 확인

- target 컬럼: {markdown_safe(target_result.get("target_column", TARGET_COLUMN))}
- target 존재 여부: {markdown_safe(target_result.get("target_exists", "확인 필요"))}
- target 결측치 수: {markdown_safe(target_result.get("target_missing_count", "확인 필요"))}
- target class 개수: {markdown_safe(target_result.get("target_unique_count", "확인 필요"))}
- target 분포:

{dict_to_bullets(target_distribution_count, "- target 분포 확인 필요")}

## 8. leakage 점검

- target과 유사한 feature 후보:

{bullet_list(target_like_columns)}

- 미래 정보 컬럼 후보:

{bullet_list(future_info_columns)}

- 식별자 컬럼 후보:

{bullet_list(id_candidate_columns)}

- 제거 또는 검토할 컬럼:

{bullet_list(review_columns)}

## 9. baseline 준비 가능 여부

- 가능 여부: {markdown_safe(baseline_ready)}
- 이유:

{bullet_list(baseline_reason_items)}

- 추가 확인 필요 사항:

{bullet_list(additional_check_items)}

## 10. 작성한 산출물

| 산출물 | 경로 |
|---|---|
| 데이터 검증 JSON | outputs/data_validation_report.json |
| 데이터 검증 Markdown | outputs/data_validation_report.md |
| 전처리 재사용 모듈 | src/preprocess.py |

## 11. OpenShift Local 연결

- OpenShift Local에서 같은 검증 로직을 실행하려면 무엇이 같아야 하는가?
- 같은 프로젝트 폴더 구조를 사용해야 한다.
- 같은 `requirements.txt`를 사용해야 한다.
- 같은 `.env.example` 구조를 유지해야 한다.
- 같은 `src/preprocess.py` 모듈을 사용할 수 있어야 한다.
- S3 bucket, S3 key, local clean file 경로를 환경 변수 또는 ConfigMap으로 주입할 수 있어야 한다.

## 12. g4dn 연결

- 이번 교시에서 g4dn을 사용하지 않는 이유는?
  - 데이터 검증은 pandas 기반 CPU 작업이므로 GPU가 필요하지 않다.

- GPU 학습 전에 데이터 검증이 필요한 이유는?
  - 잘못된 데이터로 GPU 학습을 실행하면 시간과 비용이 낭비된다.
  - target 누락, leakage, 결측치, 컬럼 불일치를 먼저 확인해야 한다.
  - 검증된 feature 데이터만 GPU 학습 단계로 넘기는 것이 안전하다.

## 13. 오늘 배운 핵심

- 데이터 검증은 모델링 전에 반드시 수행해야 한다.
- expected columns와 actual columns를 비교해야 한다.
- target 컬럼은 존재 여부, 결측치, class 개수를 확인해야 한다.
- leakage 후보 컬럼은 feature로 사용하기 전에 반드시 검토해야 한다.
- 검증 로직은 `src/preprocess.py`로 분리하면 재사용할 수 있다.

## 14. 다음 교시에서 할 일

feature 데이터 생성 + 저장 구조 정리
"""

# 앞뒤 공백 정리 후 저장
period3_record_text = textwrap.dedent(period3_record_text).strip() + "\n"

period3_record_file.write_text(
    period3_record_text,
    encoding="utf-8"
)

print("Day 5 3교시 데이터 검증 실습 기록지 저장:", period3_record_file)
print("존재 여부:", period3_record_file.exists())
print("상대 경로:", period3_record_file.relative_to(BASE_DIR))
````
